### Consolidate price-quality findings

```python
price_summary = pd.DataFrame({
    "PriceCategory" : 
    ["Negative",
    "Zero",
    "Posiitve",
    "Missing"],

    "RecordCount" : [
        (df["UnitPrice"] < 0).sum(),
        (df["UnitPrice"] == 0).sum(),
        (df["UnitPrice"] > 0).sum(),
        (df["UnitPrice"]).isna().sum()

    ],

    "LineRevenue": [
        df.loc[df["UnitPrice"] < 0, "LineRevenue"].sum(),
        df.loc[df["UnitPrice"] == 0, "LineRevenue"].sum(),
        df.loc[df["UnitPrice"] > 0, "LineRevenue"].sum(),
        df.loc[df["UnitPrice"].isna() , "LineRevenue"].sum()
    ]
})

price_summary["Precentage"] = (
    price_summary["RecordCount"] / len(df) * 100
).round(2)

display(price_summary)
```

### Investigate price and quantity together

```python
price_category = pd.Series("positive price", index=df.index)

price_category.loc[df["UnitPrice"] == 0 ] = "Zero price"
price_category.loc[df["UnitPrice"] < 0 ] = "Negative price"
price_category.loc[df["UnitPrice"].isna() ] = "Missing price"

quantitiy_category = pd.Series("positive quantity", index=df.index) 
quantitiy_category.loc[df["Quantity"] == 0] = "Zero quantity"
quantitiy_category.loc[df["Quantity"] < 0] = "Negative quantity"

pd.crosstab(price_category, quantitiy_category)
```
