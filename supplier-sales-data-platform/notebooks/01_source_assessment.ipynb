Whether invoice num beginning with C are associated with negative qunatities

```python
cancelled_invoice = df["InvoiceNo"].str.startswith("C", na=False)
negative_quantitiy = df["Quantity"] < 0

summary= pd.DataFrame({
    "Invoice starts with C" : [
        ((df["InvoiceNo"].str.startswith("C", na=False)) & (df["Quantity"] < 0)).sum(),
        ((df["InvoiceNo"].str.startswith("C", na=False)) & (df["Quantity"] >= 0)).sum()
    ],
    "Invoice does not start with C":[
        ((~df["InvoiceNo"].str.startswith("C", na=False)) & (df["Quantity"] < 0)).sum(),
        ((~df["InvoiceNo"].str.startswith("C", na=False)) & (df["Quantity"] >= 0)).sum()
    ]
}, index=["Negative quantity", "Zero or positive quantity"])

summary
```

```python
grouped = (
    df.groupby(["InvoiceNo", "StockCode"])
    .agg(
            row_count = ("Quantity", "size"),
            different_quantities = ("Quantity", "nunique")
    )
    .reset_index()
)

quantity_variation = grouped[
    (grouped["row_count"] > 1) &
    (grouped["different_quantities"] > 1)
]

quantity_variation.shape[0]
```
